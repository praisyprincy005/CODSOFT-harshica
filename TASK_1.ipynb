```python
a=int(input("Enter number:"))
b=int(input("Enter number:"))
operation= input("Enter which operation- add/sub/mul/div:  ")
if operation=="add" :
    print(a+b)
if operation=="sub":
    print(a-b)
if operation=="mul" :
    print(a*b)
if operation=="div" :
   print(a/b)
```
