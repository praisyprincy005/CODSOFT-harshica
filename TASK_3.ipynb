```python
task = []

while True:
    print(" TO-DO-LIST ")
    print(" 1. Add task ")
    print(" 2. Remove task ")
    print(" 3. View task ")
    print(" 4. Exit ")

    user = int(input(" Enter your choice: "))

    if user == 1:
        print(" 1. Add task ")
        task.append(input("Enter task:"))
        print("Task added successfully")

    elif user == 2:
        print(" 2. Remove task ")
        task.remove(input("Enter task:"))
        print("Task removed successfully")

    elif user == 3:
        print(" 3. View task ")
        print(task)

    elif user == 4:
        print(" 4. Exit ")
        break

    else:
        print("Invalid choice")
```
